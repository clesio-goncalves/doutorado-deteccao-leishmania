# Importações e parâmetros gerais

```python
import pandas as pd
import numpy as np
import random
import os
import shutil
import zipfile

from sklearn.utils import shuffle
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
# Diretorio do dataset
extensao = '.png'
diretorio_cvat = 'dataset_cvat'
subpasta_objetos = 'obj_train_data'

# Divisão do dataset
TRAIN_SIZE = 0.8
VAL_SIZE = 0.2

# Garantir que toda divisão receba as mesmas imagens
SEED = 42
np.random.seed = SEED
random.seed = SEED
```

# Descompacta o dataset baixado do CVAT

- Descompacta os arquivos e renomeia a pasta

```python
def descompacta_arquivos(diretorio_cvat):
    for index, arquivo in enumerate(os.listdir(diretorio_cvat)):
        if arquivo.endswith('.zip'):
            caminho_arquivo_zip = os.path.join(diretorio_cvat, arquivo)
            caminho_pasta_destino = os.path.join(diretorio_cvat, arquivo.split('-')[0][5:])
    
            # Extrair arquivos zip
            with zipfile.ZipFile(caminho_arquivo_zip,"r") as arquivo_zip:
                arquivo_zip.extractall(caminho_pasta_destino)
    
            # Remover o arquivo ZIP após a extração
            os.remove(caminho_arquivo_zip)
    
            print(f'{index+1} - Arquivos extraídos de {arquivo} e ZIP removido.')
```

```python
descompacta_arquivos(diretorio_cvat)
```

# Insere as imagens das lâminas em um dataframe do pandas

```python
# Pega todas as imagens de todas as lâminas
def retorna_imagens(diretorio_cvat, subpasta_objetos, extensao):
    dataset = pd.DataFrame(columns=['diretorio', 'imagem', 'lamina'])

    for lamina in os.listdir(diretorio_cvat):  # varre todas as lâminas
        diretorio = os.path.join(diretorio_cvat, lamina, subpasta_objetos)
        arquivos = os.listdir(diretorio)

        for arquivo in arquivos:  # varre todos os arquivos da lâmina
            if arquivo.endswith(extensao):
                dataset = dataset._append({'diretorio': diretorio, 'imagem': arquivo, 'lamina': lamina}, ignore_index=True)

        print(f"A lâmina {lamina} possui {len(arquivos) // 2} imagens.")

    return dataset
```

```python
dataset = retorna_imagens(diretorio_cvat, subpasta_objetos, extensao)
```

# Divide as imagens em treino e validação

- Divide as imagens mantendo uma proporção de imagens por lâmina nos datasets de treino e validação

```python
# Criando um objeto StratifiedShuffleSplit
sss = StratifiedShuffleSplit(n_splits=1, test_size=VAL_SIZE, random_state=SEED)

# Obtendo os índices para treino e teste
for train_idx, valid_idx in sss.split(dataset, dataset['lamina']):
    train_dataset = dataset.iloc[train_idx]
    valid_dataset = dataset.iloc[valid_idx]
```

# Cria o dataset no padrão da YOLO v8 com base no dataframe

## Cria as pastas e subpastas

```python
# Caminho base para o diretório principal
diretorio_yolo = "dataset_yolo"

# Subpastas
train_directory = os.path.join(diretorio_yolo, "train")
valid_directory = os.path.join(diretorio_yolo, "valid")

# Verifica se o diretório principal existe
if not os.path.exists(diretorio_yolo):
    os.makedirs(diretorio_yolo)

# Cria as subpastas train, valid e test
os.makedirs(train_directory, exist_ok=True)
os.makedirs(valid_directory, exist_ok=True)
```

## Copia as imagens e txt para os diretórios

```python
def copia_arquivos(dataset_origem, diretorio_destino):
    
    # Itera sobre o DataFrame e copia cada imagem para a pasta de destino
    for index, row in dataset_origem.iterrows():
        arquivo = os.path.splitext(row['imagem'])[0]
        
        imagem_origem = os.path.join(row['diretorio'], row['imagem'])
        txt_origem = os.path.join(row['diretorio'], f"{arquivo}.txt")

        imagem_destino = os.path.join(diretorio_destino, row['imagem'])
        txt_destino = os.path.join(diretorio_destino, f"{arquivo}.txt")

        # Copia a imagem e txt
        shutil.copy(imagem_origem, imagem_destino)
        shutil.copy(txt_origem, txt_destino)

    print(f"Cópia de {dataset_origem.shape[0] * 2} arquivos concluída.")
```

```python
# Dataset de treino
copia_arquivos(train_dataset, train_directory)
```

```python
# Dataset de validação
copia_arquivos(valid_dataset, valid_directory)
```

## Adiciona nome e caminho de todas as imagens em um txt

```python
def adiciona_caminhos_imagens(dataset, diretorio):

    global diretorio_yolo

    nome_arquivo = diretorio.split("/")[-1] + ".txt"
    
    os.chdir(os.path.join(diretorio_yolo))
    with open(nome_arquivo, "w") as outfile:
        for index, row in dataset.iterrows():
            caminho_imagem = os.path.join(diretorio, row['imagem'])
    
            outfile.write(caminho_imagem)
            outfile.write("\n")
        outfile.close()
    os.chdir("..")

    print(f"Arquivo {nome_arquivo} criado com sucesso")
```

```python
adiciona_caminhos_imagens(train_dataset, train_directory)
```

```python
adiciona_caminhos_imagens(valid_dataset, valid_directory)
```

## Cria os arquivos obj.data e obj.names

```python
# Copia o obj.names
obj_names = os.path.join(diretorio_cvat, valid_dataset['lamina'].iloc[0], "obj.names")
shutil.copy(obj_names, diretorio_yolo)
```

```python
# Pega o número de classes do arquivo obj.names
NUM_CLASSES = sum(1 for _ in open(obj_names))
```

```python
# Cria o arquivo obj.data
obj_data = os.path.join(diretorio_yolo, "obj.data")
diretorio_obj_names = os.path.join(diretorio_yolo, "obj.names")
with open(obj_data, 'w') as arquivo:
    arquivo.write(f"classes = {NUM_CLASSES}\n")
    arquivo.write(f"train = {train_directory}\n")
    arquivo.write(f"valid = {valid_directory}\n")
    arquivo.write(f"names = {diretorio_obj_names}\n")
    arquivo.write("backup = yolov8/recursos/\n")
```
